```python
import sys
from pathlib import Path

ROOT = Path(r"C:\Users\DELL\OneDrive\Documents\Anurag\Anurag_Projects\INTERNMO\Netflix Customer Intelligence & AI Analytics Copilot")

assert (ROOT / "src" / "utils" / "config.py").exists(), f"config.py not found under {ROOT}"
sys.path.insert(0, str(ROOT))
print("Project root:", ROOT)

import pandas as pd
import numpy as np
from src.utils.config import raw_path, RAW_FILES

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)

tables = {name: pd.read_csv(raw_path(name)) for name in RAW_FILES}
for name, df in tables.items():
    print(f"{name:20s} {df.shape}")
```

```python
def profile(name, df):
    print(f"\n{'='*70}\n{name.upper()}  shape={df.shape}  full-row duplicates={df.duplicated().sum()}\n{'='*70}")
    out = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "nulls": df.isna().sum(),
        "null_%": (df.isna().mean() * 100).round(2),
        "unique": df.nunique(),
    })
    display(out)

for name, df in tables.items():
    profile(name, df)
```

```python
cust = tables["customers"]
print("Duplicate customer_id rows:", cust.customer_id.duplicated().sum())
print("Unique customers:", cust.customer_id.nunique())
print("\nCountry values:\n", cust.country.value_counts(dropna=False).head(40))
print("\nAge describe:\n", cust.age.describe())
print("Ages <10 or >100:\n", cust[(cust.age < 10) | (cust.age > 100)][["customer_id", "age"]])

va = tables["viewing_activity"]
print("\nDevice values:\n", va.device_type.value_counts(dropna=False))
print("\nLogin location values:\n", va.login_location.value_counts(dropna=False).head(40))
print("\nWatch minutes describe:\n", va.watch_duration_minutes.describe())

st = tables["support_tickets"]
parsed = pd.to_datetime(st.ticket_date, errors="coerce")
print("\nInvalid ticket_date rows:", parsed.isna().sum())
print(st.loc[parsed.isna(), "ticket_date"].value_counts().head(10))

pay = tables["payments"]
print("\nPayment amount describe:\n", pay.amount.describe())
print("payments.subscription_id non-null:", pay.subscription_id.notna().sum())

print("\nChurn customers not in customers table:",
      (~tables["churn_labels"].customer_id.isin(cust.customer_id)).sum())
```

```python
sub = tables["subscriptions"]; st = tables["support_tickets"]; ch = tables["churn_labels"]

print("ALL customer countries:\n", sorted(cust.country.dropna().unique()))
print("\nlogin_location values NOT in customers.country:\n",
      sorted(set(va.login_location.dropna()) - set(cust.country.dropna())))

print("\nPlan prices:\n", tables["subscription_plans"][["plan_id","plan_name","monthly_price"]])
print("\nPayments by status and amount:\n", pay.groupby(["payment_status","amount"]).size())

print("\nSubscription status vs (has end_date, has cancel_date):")
print(pd.crosstab(sub.subscription_status, [sub.subscription_end_date.notna(), sub.cancellation_date.notna()]))

print("\nWatch minutes > 300:", (va.watch_duration_minutes > 300).sum(),
      "| watch > session:", (va.watch_duration_minutes > va.session_duration_minutes).sum())
print(va.watch_duration_minutes.quantile([.99, .999]))

print("\nTicket status vs CSAT missing:")
print(pd.crosstab(st.ticket_status, st.customer_satisfaction_score.isna()))

print("\nChurn:", ch.churned.value_counts().to_dict(), "| rate:", round(ch.churned.mean(), 4))
```
